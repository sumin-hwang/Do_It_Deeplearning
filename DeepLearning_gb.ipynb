# 04. 선형회귀
- 임의의 직선을 그어 이에 대한 평균 제곱 오차를 구하고, 이 값을 가장 작게 만들어 주는 a와 b를 찾아가는 작업

- 딥러닝을 설계한다는것 : 몇 개의 층을 어떻게 싸을지, 어떤 층을 사용할지, 내부의 변수를 어떻게 정해야 하는지..


```python
# 파이썬 기초
import numpy as np
import pandas as pd

x = np.array([2,4,6,8])
y = np.array([81, 93, 91, 97])

mx = np.mean(x)
my = np.mean(y)

print("mx", mx)
print("my", my)

# 분모
divisor = sum([(i - mx)**2 for i in x])

# 분자
def top(x, mx, y, my):
  d = 0
  for i in range(len(x)):
    d += (x[i] - mx)* (y[i] - my)
  return d

dividend = top(x, mx, y, my)

# 출력
print("mo", divisor)
print("za", dividend)

# a
a = dividend / divisor

# b
b = my - (mx * a)

print("a", a)
print("b", b)
```

```python
import numpy as np

# 선형회귀

fake_a = 3
fake_b = 76
predict_result = []

x = np.array([2,4,6,8])
y = np.array([81, 93, 91, 97])


def predict(x):
  return fake_a * x + fake_b

for i in range(len(x)):
  predict_result.append(predict(x[i]))
  print("공부시간=%.f, 실제점수=%.f, 예측점수=%.f" %(x[i], y[i], predict(x[i])))

n = len(x)
def mse(y, y_pred):
  return (1/n)* sum ((y - y_pred)**2)

print(str(mse(y, predict_result)))
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([2,4,6,8])
y = np.array([81, 93, 91, 97])

# 데이터 분포
plt.scatter(x, y)
plt.show()

# 기울기 a,b 초기화
a = 0
b = 0

# 학습률
lr = 0.03

# 반복횟수
epochs = 2001

# x값이 총 몇개인지
n = len(x)

# 경사하강법
for i in range(epochs):
  y_pred = a*x + b
  error = y - y_pred # 오차 error

  a_diff = (2/n)*sum(-x * (error)) # 편미분값, 각 변수를 조금씩 바꿨을 때 어떻게 손실이 바꿨는지 알기 위해 편미분한다. -> 기울기 a를 바꿀 때의 손실
  b_diff = (2/n)* sum (-error)  # 절편 b를 바꿀 때의 손실

  a = a - lr * a_diff
  b = b - lr * b_diff # 학습률을 곱해서 기존의 b값을 업데이트함

  if i %100 == 0:
    print("epoch=%.f, 기울기 a=%.4f, 절편 b=%.4f" %(i, a, b))

# 최종 a, b
y_pred = a*x + b

plt.scatter(x, y)
plt.plot(x, y_pred, 'r')
plt.show()





```

```python
# 다중선형회귀

x1 = np.array([2,4,6,8])
x2 = np.array([0, 4, 2, 3])
y = np.array([81, 93, 91, 97])

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter3D(x1, x2, y)
plt.show()

a1 = 0
a2 = 0
b = 0

lr = 0.01
epochs = 2001

n = len(x1)

# 경사하강법
for i in range(epochs):
  y_pred = a1 * x1 + a2 * x2 + b # 예측값
  error = y - y_pred # 오차

  a1_diff = (2/n)* sum(-x1 * error)
  a2_diff = (2/n)* sum (-x2 * error)
  b_diff = (2/n)* sum(-(error)) # 오차 함수를 b로 편미분한 값

  # 학습률 곱해 업데이트
  a1 = a1 - lr * a1_diff
  a2 = a2 - lr * a2_diff
  b = b- lr * b_diff

  if i %100 == 0:
    print("epoch=%.f, 기울기1=%.04f, 기울기2=%.04f, 절편=%.04f" %(i, a1, a2, b))

print("실제 점수 : ", y)
print("예측 점수: ", y_pred)
```
