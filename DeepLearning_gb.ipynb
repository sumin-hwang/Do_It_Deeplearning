# 04. 선형회귀

```python
# 파이썬 기초
import numpy as np
import pandas as pd

x = np.array([2,4,6,8])
y = np.array([81, 93, 91, 97])

mx = np.mean(x)
my = np.mean(y)

print("mx", mx)
print("my", my)

# 분모
divisor = sum([(i - mx)**2 for i in x])

# 분자
def top(x, mx, y, my):
  d = 0
  for i in range(len(x)):
    d += (x[i] - mx)* (y[i] - my)
  return d

dividend = top(x, mx, y, my)

# 출력
print("mo", divisor)
print("za", dividend)

# a
a = dividend / divisor

# b
b = my - (mx * a)

print("a", a)
print("b", b)
```
